# What a DataFrame is?

A DataFrame is a table, it is like an Excel sheet.
Each **column** has a **name and one type**. Each **row** has an **index**(it'a label, starting at 0)

```python
import pandas as pd

df = pd.DataFrame ({
    "name" : ["Zohaina", "Kevin", "Hery"],
    "group" : ["A", "B", "C"],
    "Age" : [22, 22, 22],
    "Country" : ["China", "Indonesia", "Madagascar"],
})

print(df)
print(df[["name", "group"]])
print()
print(df.shape)
print()
print(df.head(2))
print()
print(df.tail(1))
print()
print(df.dtypes)
print(df.columns)
print(df.describe)
print(df.info)
```

### Selecting Columns

```python
df["Age"]  #= df.Age
df[["Age", "name"]]
df["Age"].mean()
```
