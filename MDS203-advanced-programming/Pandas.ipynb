# What a DataFrame is?

A DataFrame is a table, it is like an Excel sheet.
Each **column** has a **name and one type**. Each **row** has an **index**(it'a label, starting at 0)

```python
import pandas as pd

df = pd.DataFrame ({
    "name" : ["Zohaina", "Kevin", "Hery"],
    "group" : ["A", "B", "A"],
    "Age" : [22, 23, 22],
    "score" : [89, 65, 73],
    "Country" : ["China", "Indonesia", "Madagascar"],
})

print(df)
print(df[["name", "group"]])
print()
print(df.shape)
print()
print(df.head(2))
print()
print(df.tail(1))
print()
print(df.dtypes)
print(df.columns)
print(df.describe)
print(df.info)
```

### Selecting Columns

```python
df["Age"]  #= df.Age
df[["Age", "name"]]
df["Age"].mean()
```

### `loc` vs `iloc`

```python
# df = pd.DataFrame({
#     "descri" : ["uses", "slice end","memory trick"],
#     "loc" : ["labels", "included", "Loc=label"],
#     "iloc" : ["positions", "exluded", "iloc=integer position"],
# })

# df[["descri", "loc", "iloc"]]
```

```python
df.loc[1]                       # the row with index label 1 → Budi
```

```python
df.iloc[0]                      # the row in position 0 → Ani
```

```python
df.iloc[0, 2]                   # row 0, column 2 (age) → 21
```

```python

df.loc[2, "score"]              # row label 2, column "score" → 72
```

```python

df.loc[1:2, ["name", "Age"]]    # rows 1 AND 2 (end included)
```

```python

df.iloc[1:2]               # only row 1 (end excluded, like range)
```

Same `1:2`, different results. That's a typical MCQ question.

Here the labels and positions are the same numbers (0–3). After you sort or delete rows they stop matching, and then mixing up `loc` and `iloc` gives wrong rows.

#### `groupby`

```python
df.head(3)
```

`groupby` = split the rows into groups → compute something in each group → combine the results

```python
df.groupby("group")["score"].mean()
```

```python
df.groupby("group")["score"].agg(["mean", "max", "min","count" ])
```

RangeData = Custom Aggregate Function

```python
def RangeData(x):
    return x.max() - x.min()
```

```python
df.groupby("group")["score"].agg(RangeData)
```
