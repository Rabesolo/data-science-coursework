# Python Coursework

## Section 1: Types

- `int(3)`
- `float(1.5)`
- `str('Hello')`
- `bool(True)`
- `None`
You can check the type of your input with **type(x)**.
**Floats** are approximate so `0.1+0.3==0.4` is `False`.
`math.isclose()` is to check either the variables are close to each other or not.

## Section 2: Collections

- `list[1, 2]` :oredered and changeable
- `tuple(1, 2)` :unchangeable
- `set{1, 2}` :no duplicates
- `dict{'a': 1}` : key==>value

## Section 3: `for` loops

- `range(start, stop, step)`
- `enumerate()` :gives index and value. (ex: 0 Blue, 1 Red, 2 Green, ...)
- `zip()` : combine multiple iterables. (zip(names, score))
- `break` : stops immediatly the loop.
- `continue` : skips to the next item.
- `pass` : Does nothing.

## Section 4: `if/elif/else` 

Python checks top to bottom and runs only the first true branch, so put specific conditions before general ones.

## Section 5: `while` and exceptions

- `while` repeats while a condition is true, so something must change or it never stops.
- `try / except ValueError:` catches error such as `int('abc')`. 

## Section 6: Functions
 - `def name(param, default=5):`defines one function.
 - `return` :sends back a result.
 - A function without `return` gives **`None`**
