## Concept 2: Loops and Accumulation
Given scores = `[80, 95, 70, 88]`, compute the average without using sum(), with a for loop.

```python
scores = [80, 95, 70, 88]
total = 0
count = 0

for score in scores:
    total += score
    count += 1
    
average = total/count
print(count, total, average)
```

## Concept 3: `enumerate` and `zip`
Print `1. Ana: 78`, `2. Budi: 85`, `3. Kevin: 92` with one loop.

```python
names = ['Ana', 'Budi', 'Kevin']
marks = [78, 85, 92]

for name, mark in zip(names,marks):
    print(name, mark)

for position, name in enumerate(names, start=1):
    print(position, name)

for position, (name, mark) in enumerate(zip(names, marks), start=1):
    print(f"{position}. {name}= {mark}")
```
